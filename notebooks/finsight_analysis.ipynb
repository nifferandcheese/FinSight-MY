```python
income = pd.read_csv("../data/raw/hh_income_district.csv")
poverty = pd.read_csv("../data/raw/hh_poverty_district.csv")
inequality = pd.read_csv("../data/raw/hh_inequality_district.csv")

print("Income:", income.shape)
print("Poverty:", poverty.shape)
print("Inequality:", inequality.shape)
```

```python
#MERGE THE DATASETS 
income_df = income.copy()
poverty_df = poverty.copy()
inequality_df = inequality.copy()

# Convert dates to datetime
income_df["date"] = pd.to_datetime(income_df["date"])
poverty_df["date"] = pd.to_datetime(poverty_df["date"])
inequality_df["date"] = pd.to_datetime(inequality_df["date"])

# Merge income + poverty
merged_df = pd.merge(
    income_df,
    poverty_df,
    on=["state", "district", "date"],
    how="inner"
)

# Merge with inequality
merged_df = pd.merge(
    merged_df,
    inequality_df,
    on=["state", "district", "date"],
    how="inner"
)

print("Merged dataset shape:", merged_df.shape)
display(merged_df.head())
```

```python
print("Missing values:")
print(merged_df.isnull().sum())

print("\nDuplicate rows:")
print(merged_df.duplicated().sum())
```

```python
print("\nYears:")
print(sorted(merged_df["date"].dt.year.unique()))

print("\nNumber of districts:")
print(merged_df["district"].nunique())
```

```python
merged_df.to_csv(
    "../data/processed/finsight_merged.csv",
    index=False
)

print("Merged dataset saved successfully!")
```

```python
#CREATE A FUTURE FINANCIAL RISK TARGET 
# Sort by district and year
merged_df = merged_df.sort_values(["district", "date"])

# Calculate change in relative poverty for each district
merged_df["future_poverty"] = merged_df.groupby("district")["poverty_relative"].shift(-1)

merged_df["poverty_change"] = (
    merged_df["future_poverty"] - merged_df["poverty_relative"]
)

display(
    merged_df[
        ["state", "district", "date", "poverty_relative",
         "future_poverty", "poverty_change"]
    ].head(15)
)
```

```python
print(merged_df["poverty_change"].describe())
```

```python
#CHECK THE RISK CLASSES 
risk_df = merged_df.dropna(subset=["poverty_change"]).copy()

risk_df["financial_risk"] = pd.cut(
    risk_df["poverty_change"],
    bins=[-float("inf"), 0, 3, float("inf")],
    labels=["Low", "Medium", "High"],
    right=False
)

print("Risk distribution:")
print(risk_df["financial_risk"].value_counts())
```

```python
print("\nRisk percentage:")
print(
    risk_df["financial_risk"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Create final ML dataset
ml_df = risk_df.copy()

# Create numeric target
ml_df["financial_risk"] = ml_df["financial_risk"].map({
    "Low": 0,
    "Medium": 1,
    "High": 2
})

# Select features and target
ml_df = ml_df[
    [
        "state",
        "district",
        "date",
        "income_mean",
        "income_median",
        "poverty_absolute",
        "poverty_relative",
        "gini",
        "financial_risk"
    ]
]

print("ML dataset shape:", ml_df.shape)
display(ml_df.head())
```
